```python
import numpy as np
```

```python
import time
```

```python
shape = 5000
```

```python
np_array = np.random.uniform(size = [shape, shape])
```

```python
start = time.time()

for i in range(10):
    np.matmul(np_array, np_array)

end = time.time()
time_period = end - start
print(f'{time_period} seconds')
```

```python
tf_array = tf.random.uniform([shape, shape])
```

```python
start = time.time()

for i in range(10):
    tf.matmul(tf_array, tf_array)

end = time.time()
time_period = end - start
print(f'{time_period} seconds')
```
